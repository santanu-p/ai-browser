```python
# 3) Run a crawl and save JSON output
import subprocess
cmd = [
    'agentic-browser', 'https://example.com',
    '--max-pages', '10', '--max-depth', '1', '--concurrency', '5', '--out', 'out.json'
]
result = subprocess.run(cmd, text=True, capture_output=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(f'Crawl failed with exit code {result.returncode}')
```

```python
# 4) Inspect crawl output
import json
with open('out.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
print('Pages crawled:', len(data.get('pages', [])))
print('Top-level keys:', list(data.keys()))
```
